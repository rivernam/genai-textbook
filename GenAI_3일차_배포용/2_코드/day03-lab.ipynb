### 환경 준비 (1) 실습 재료와 라이브러리

**실습 재료를 내려받고 라이브러리를 설치합니다. 노트북을 열 때마다 가장 먼저 한 번 실행합니다.**

1. 코랩이면 학습 사이트에서 실습 자료 zip 두 개(`day03-examples.zip` · `day03-missions.zip`)를 내려받아 풉니다. 이미 파일 창에 올려 두었으면 그 파일을 씁니다.
2. 코랩에 없거나 버전을 맞출 라이브러리(`anthropic` · `sentence-transformers` · `python-docx`)와 PDF 를 텍스트로 바꾸는 `pdftotext`(poppler-utils)를 설치합니다.

```python
# 환경 준비: 노트북을 열면 이 셀부터 한 번 실행합니다
# 코랩에서는 실습 자료 zip 두 개를 내려받아 풀고, 없는 라이브러리만 설치합니다.
import os, sys, glob, subprocess

IN_COLAB = 'google.colab' in sys.modules
SITE = 'https://lecture-web-five.vercel.app/materials/'

if IN_COLAB:
    os.chdir('/content')
    for z in ['day03-examples.zip', 'day03-missions.zip']:
        if os.path.exists(z) and os.path.getsize(z) == 0:
            os.remove(z)                                # 지난번 내려받기가 실패해 남은 빈 파일
        if not os.path.exists(z):                       # 파일 창에 직접 올렸으면 그대로 씁니다
            if subprocess.run(['wget', '-q', SITE + z, '-O', z]).returncode != 0:
                if os.path.exists(z):
                    os.remove(z)
                raise SystemExit(f'{z} 를 내려받지 못했습니다. 학습 사이트에서 받아 왼쪽 파일 창에 올린 뒤 이 셀을 다시 실행하세요.')
        if subprocess.run(['unzip', '-q', '-o', z]).returncode != 0:
            os.remove(z)                                # 풀리지 않는 파일은 지우고 다음 실행에서 다시 받습니다
            raise SystemExit(f'{z} 가 풀리지 않아 지웠습니다. 이 셀을 다시 실행하세요.')
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q',
                    'anthropic', 'sentence-transformers', 'python-docx'])
    subprocess.run('apt-get -qq install -y poppler-utils > /dev/null', shell=True)

import numpy, sentence_transformers
print('numpy', numpy.__version__, '· sentence-transformers', sentence_transformers.__version__)
print('문서', len(os.listdir('corpus')), '건 · 원본 문서',
      len(glob.glob('docs_raw/*/*')), '건 · 미션 재료',
      len(glob.glob('3일차_미션*_재료.txt')), '개')
```

### 미션 1 준비 · 문서 253건 인덱스를 다시 만듭니다
**문서 253건으로 인덱스를 다시 만드는 셀입니다. 실행만 합니다.**

예제 1-1 (a) 와 같은 셀입니다. 문서 253건을 읽고 임베딩 모델로 인덱스 `E` 를 만듭니다. API 를 호출하지 않습니다.
미션 셀 안내의 「예제 1-1 (a)」는 이 노트북에서는 이 셀을 가리킵니다. 실습 미션 1 (b) 를 처음부터 다시 실행할 때도 이 셀부터 실행합니다.

**볼 것** `253 건 · 인덱스 (253, 384)` 가 나오면 미션 1 (a) 로 넘어갑니다.

```python
# 미션 1 준비 · 예제 1-1 (a) 와 같은 셀: 문서 253건 인덱스를 다시 만듭니다 (문서 253건 → 벡터 253개)
import os, json, unicodedata
import numpy as np
from sentence_transformers import SentenceTransformer

files = sorted(f for f in os.listdir('corpus') if f.endswith('.md'))
names = [unicodedata.normalize('NFC', f[:-3]) for f in files]     # 맥에서 푼 한글 파일명 대비
docs  = [open(f'corpus/{f}', encoding='utf-8').read() for f in files]

model = SentenceTransformer('intfloat/multilingual-e5-small')
E = model.encode(['passage: ' + d for d in docs],
                 normalize_embeddings=True, batch_size=32)
print(len(docs), '건 · 인덱스', E.shape, '·', round(E.nbytes / 1e6, 3), 'MB')
```

```python
# 실습 미션 1 (a) 질문 6개를 반영 전 인덱스로 먼저 측정합니다
# 볼 것: 「인덱스 (253, 384)」와 질문 6개의 반영 전 순위. 아직 없는 문서를 묻는 질문은 「없음」이 정상입니다
def embed(text):                       # 문서 한 건을 벡터 하나로 (예제 1-1 (f) 와 같은 계산)
    return model.encode(['passage: ' + text], normalize_embeddings=True)[0]

def read(nm):                          # 미션1_변경문서/ 에서 개정본 · 신규 문서를 읽습니다
    return open(f'미션1_변경문서/{nm}.md', encoding='utf-8').read()

def top(q, k):                         # 질문 하나의 상위 k 건 문서 이름
    v = model.encode(['query: ' + q], normalize_embeddings=True)[0]
    return [names[t] for t in (E @ v).argsort()[::-1][:k]]

MYSET = [('에치 시즈닝 런은 더미 웨이퍼 몇 장 돌리나요?',       'SOP-ETCH-005'),
         ('가스 실린더 바꾸고 감지기는 몇 분 동안 봐야 하나요?', 'EHS-001'),
         ('ETCH-03 PM 요일이 바뀐다던데 무슨 요일이에요?',       'NOTI-101'),
         ('챔버 웻 클리닝은 어떤 순서로 하나요?',                'SOP-ETCH-020'),
         ('에치 웨이퍼를 손으로 넣고 뺄 때 순서',                'SOP-ETCH-001'),
         ('연말 셧다운 일정 알려주세요',                          None)]

def ranks_of(qset):                    # 질문마다 정답 문서의 순위 (없으면 None)
    out = []
    for q, gold in qset:
        order = top(q, len(names))
        r = order.index(gold) + 1 if gold in order else None
        out.append(r)
        rank = f'{r}위' if r else '없음'
        print(f'{q[:24]:26s} 정답 {rank} · 1위 {order[0]}')
    return out

print('인덱스', E.shape)
before = ranks_of(MYSET)

# 반영 전 인덱스의 사본을 보관합니다. 바꿔 보기 셀은 이 사본으로 계산하고 E · names · docs 는 바꾸지 않습니다
if len(names) == 253:
    E0, names0, docs0 = E.copy(), list(names), list(docs)
```

```python
# 실습 미션 1 (b) 변경 목록을 인덱스에 반영합니다
# 볼 것: 「반영 후 인덱스 (250, 384) · 이름표 250 · 기대 250」

# 반영할 개정 목록 (재료 [1] 의 1 · 2번)
REVISED = ['SOP-ETCH-005', 'EHS-001']
# 지울 폐기 목록 (재료 [1] 의 5번)
RETIRED = ['NOTI-001', 'NOTI-007', 'NOTI-013', 'NOTI-019', 'NOTI-025']

ADDED = ['NOTI-101', 'SOP-ETCH-020']       # 신규 목록 (재료 [1] 의 3 · 4번)
assert len(names) == 253 and not any(nm in names for nm in ADDED), \
    '(b) 를 이미 실행했습니다. 「미션 1 준비」 셀부터 다시 실행하세요'

# 개정: 바뀐 본문으로 그 문서의 행만 다시 계산합니다 (예제 1-1 (f))
for nm in REVISED:
    i = names.index(nm)
    docs[i] = read(nm)
    E[i] = embed(docs[i])

# 신규: 맨 아래에 행을 붙이고, 이름표와 본문도 같은 순서로 붙입니다 (예제 1-1 (h))
for nm in ADDED:
    E = np.vstack([E, embed(read(nm))])
    names.append(nm)
    docs.append(read(nm))

# 폐기: 같은 자리 j 를 E · names · docs 에서 함께 지웁니다 (예제 1-1 (h))
for nm in RETIRED:
    j = names.index(nm)
    E = np.delete(E, j, axis=0)          # axis=0 은 행(문서 한 건)을 지웁니다
    del names[j]
    del docs[j]

print('개정', len(REVISED), '· 신규', len(ADDED), '· 폐기', len(RETIRED))
print('반영 후 인덱스', E.shape, '· 이름표', len(names),
      '· 기대', 253 + 2 - 5)      # 변경 목록(재료 [1]) 기준: 신규 2 · 폐기 5
```

```python
# 실습 미션 1 (c) 저장 → 다시 불러오기 → 다시 측정하기
# 볼 것: 질문 2 · 3 · 4 가 1위로 올라왔는지, 6번 질문의 상위 3건에 폐기한 공지가 없는지
np.save('index.npy', E)
json.dump(names, open('index_names.json', 'w', encoding='utf-8'), ensure_ascii=False)
E = np.load('index.npy')
names = json.load(open('index_names.json', encoding='utf-8'))
# 벡터 수와 이름표 수가 다르면 멈춥니다. 그때는 「미션 1 준비」 셀부터 다시 실행합니다
assert len(E) == len(names), '벡터 수와 이름표 수가 다릅니다'
print('불러온 인덱스', E.shape, '· 이름표', len(names), '· 일치')

after = ranks_of(MYSET)
gold = [r for r, (q, g) in zip(after, MYSET) if g]          # 정답 문서가 있는 5문항
for k in [1, 3, 5]:
    print(f'recall@{k}', round(sum(r is not None and r <= k for r in gold) / len(gold) * 100), '%')

# ✏️ ======== 더 해 보기(선택) · 6번 질문에서 출력할 문서 수 ========
# 지금은 예제 1-1 (c) 에서 정한 3 이 들어 있어, 6번 질문의 상위 3건을 출력합니다.
# 1~20 사이 값(예: 10)으로 고쳐, 답이 없는 질문에도 검색이 K건을 돌려주는지와 폐기한 공지가 남았는지 확인합니다.
K = 3
# ✏️ ======== 여기까지 ========
print(f'6번 질문의 상위 {K}건:', top('연말 셧다운 일정 알려주세요', K))

# 질문 2 의 1위 문서에서 「감지기」가 들어간 줄을 찾습니다. 답의 근거가 되는 본문입니다
first = top(MYSET[1][0], 1)[0]
lines = [l for l in docs[names.index(first)].split('\n') if '감지기' in l]
print('질문 2 의 1위', first, '· 감지기 문장:', lines if lines else '없음')

# 1번처럼 순위가 그대로인 질문은 본문이 바뀌었는지 봅니다(「15장」이 보이면 개정이 들어간 것)
print(docs[names.index('SOP-ETCH-005')][:300])
```

### 미션 1 · 바꿔 보고 답하기
**바꿔 보기 1, 2 는 강사 화면을 보며 함께 실행합니다. 셀마다 ✏️ 줄 사이의 값을 고쳐 실행하고, 결과를 보고 Q1, Q2 를 적습니다.**

바꿔 보기 셀은 반영 전 인덱스의 사본에 변경 목록을 반영하고 측정해, 기본 반영(실습 미션 1 (b) 와 같은 목록)의 결과와 나란히 출력합니다. 값을 되돌리지 않아도 다른 셀의 결과는 바뀌지 않습니다.

**바꿔 보기 도우미** 아래 도우미 셀을 먼저 한 번 실행합니다.

**바꿔 보기 1** 바꿔 보기 1 셀의 개정 문서 목록 `REVISED_1` 에서 `'EHS-001'` 을 빼 `REVISED_1 = ['SOP-ETCH-005']` 로 고치고 그 셀을 실행합니다. 질문 2 의 순위 줄과 「질문 2 의 1위 · 감지기 문장」 줄을 기본 반영과 비교합니다.

**바꿔 보기 2** 바꿔 보기 2 셀의 폐기 공지 목록 `RETIRED_2` 를 두 건만 남겨 `RETIRED_2 = ['NOTI-001', 'NOTI-007']` 로 고치고 그 셀을 실행합니다. 「6번 질문의 상위 3건」 줄을 기본 반영과 비교합니다.

```python
# 실습 미션 1 바꿔 보기 도우미 · 반영 전 인덱스의 사본에 변경 목록을 반영하고 측정하는 함수입니다 · 실행만 합니다
# (a) 에서 보관한 사본(E0 · names0 · docs0)으로 계산하므로 E · names · docs 와 (b) · (c) 의 결과는 바뀌지 않습니다
assert 'E0' in globals(), '실습 미션 1 (a) 셀을 먼저 실행하세요'
BASE_REVISED = ['SOP-ETCH-005', 'EHS-001']                                   # 실습 미션 1 (b) 의 개정 목록
BASE_RETIRED = ['NOTI-001', 'NOTI-007', 'NOTI-013', 'NOTI-019', 'NOTI-025']   # 실습 미션 1 (b) 의 폐기 목록
BASE_ADDED = ['NOTI-101', 'SOP-ETCH-020']                                    # 실습 미션 1 (b) 의 신규 목록

def apply_changes(revised, retired, added=BASE_ADDED):
    E1, names1, docs1 = E0.copy(), list(names0), list(docs0)    # 사본에만 반영합니다
    for nm in revised:                     # 개정: 그 문서의 행만 다시 계산 (실습 미션 1 (b) 와 같은 순서)
        i = names1.index(nm)
        docs1[i] = read(nm)
        E1[i] = embed(docs1[i])
    for nm in added:                       # 신규: 맨 아래에 붙입니다
        E1 = np.vstack([E1, embed(read(nm))])
        names1.append(nm)
        docs1.append(read(nm))
    for nm in retired:                     # 폐기: 같은 자리를 세 곳에서 함께 지웁니다
        j = names1.index(nm)
        E1 = np.delete(E1, j, axis=0)
        del names1[j]
        del docs1[j]
    return E1, names1, docs1

def measure(label, E1, names1, docs1):    # 실습 미션 1 (b) · (c) 와 같은 줄을 출력합니다
    def top_in(q, k):
        v = model.encode(['query: ' + q], normalize_embeddings=True)[0]
        return [names1[t] for t in (E1 @ v).argsort()[::-1][:k]]
    print(f'[{label}]')
    print('  반영 후 인덱스', E1.shape, '· 이름표', len(names1), '· 기대', 253 + 2 - 5)
    for q, gold in MYSET:
        ranking = top_in(q, len(names1))
        r = ranking.index(gold) + 1 if gold in ranking else None
        rank = f'{r}위' if r else '없음'
        print(f'  {q[:24]:26s} 정답 {rank} · 1위 {ranking[0]}')
    print('  6번 질문의 상위 3건:', top_in('연말 셧다운 일정 알려주세요', 3))
    first = top_in(MYSET[1][0], 1)[0]
    lines = [l for l in docs1[names1.index(first)].split('\n') if '감지기' in l]
    print('  질문 2 의 1위', first, '· 감지기 문장:', lines if lines else '없음')

print('바꿔 보기 도우미 준비 완료')
```

```python
# 실습 미션 1 바꿔 보기 1 · 개정 목록만 바꿔 반영하고, 기본 반영과 나란히 측정합니다
# 볼 것: 질문 2 의 순위 줄과 「질문 2 의 1위 · 감지기 문장」 줄이 기본 반영과 어떻게 다른지
# ✏️ ======== 직접 고칠 곳 · 이번에 반영할 개정 문서 ========
# 지금은 실습 미션 1 (b) 에서 반영한 개정 문서 두 건이 들어 있습니다.
# 'EHS-001' 을 빼 ['SOP-ETCH-005'] 로 고쳐, 개정본 EHS-001 을 반영하지 않으면 질문 2 의 1위 문서가 어떻게 되는지 확인합니다.
REVISED_1 = ['SOP-ETCH-005', 'EHS-001']
# ✏️ ======== 여기까지 ========
measure('기본 반영 · 실습 미션 1 (b) 와 같은 목록', *apply_changes(BASE_REVISED, BASE_RETIRED))
measure(f'바꿔 보기 1 · 개정 {REVISED_1}', *apply_changes(REVISED_1, BASE_RETIRED))
```

```python
# 실습 미션 1 바꿔 보기 2 · 폐기 목록만 바꿔 반영하고, 기본 반영과 나란히 측정합니다
# 볼 것: 「6번 질문의 상위 3건」 줄과 「반영 후 인덱스 · 기대」 줄이 기본 반영과 어떻게 다른지
# ✏️ ======== 직접 고칠 곳 · 이번에 지울 폐기 공지 ========
# 지금은 실습 미션 1 (b) 에서 지운 공지 다섯 건이 들어 있습니다.
# ['NOTI-001', 'NOTI-007'] 두 건으로 고쳐, 폐기 공지를 일부만 지우면 검색 결과가 어떻게 되는지 확인합니다.
RETIRED_2 = ['NOTI-001', 'NOTI-007', 'NOTI-013', 'NOTI-019', 'NOTI-025']
# ✏️ ======== 여기까지 ========
measure('기본 반영 · 실습 미션 1 (b) 와 같은 목록', *apply_changes(BASE_REVISED, BASE_RETIRED))
measure(f'바꿔 보기 2 · 폐기 {len(RETIRED_2)}건', *apply_changes(BASE_REVISED, RETIRED_2))
```

## 실습 미션 2 · 새 원본 10건을 검색되는 청크로 자르기
**새 원본 문서 10건을 텍스트로 열고, 노이즈 줄을 지운 뒤 잘라서 검색 적중률을 측정하고 꼬리표를 붙입니다.**

### 미션 2 준비 · 예제 2-1 ~ 2-5 의 값과 함수를 다시 만듭니다
**예제 파트 2 의 값과 함수를 다시 만드는 셀입니다. 아래 두 셀 모두 실행만 합니다.**

(1) 셀은 예제 노트북 파트 2 의 준비 셀과 같습니다(문서 253건 · 임베딩 모델 · 질문 10개). (2) 셀은 예제 2-1 ~ 2-5 에서 만든 도구와 함수를 한 셀에 모았습니다. 코드는 예제와 같고, API 를 호출하지 않습니다.
- `subprocess` 예제 2-1 (b) · `DROP` 예제 2-1 (c) · `Document` 예제 2-2 (d) · `load_workbook` 예제 2-2 (f) · `pd` 예제 2-2 (g)
- `chunk_fixed()` 예제 2-3 (a) · `chunk_sent()` 예제 2-3 (c) · `chunk_struct()` 예제 2-3 (e)
- `QE` · `build_chunks()` · `evaluate()` 예제 2-4 (a) · `dept_of()` 예제 2-5 (d)

**볼 것** (2) 셀에 `준비 완료` 가 나오면 실습 미션 2 (a) 로 넘어갑니다.

```python
# 준비: 문서 253건 · 임베딩 모델 · 질문 10개 (예제 1-1 · 1-2 와 같은 값)
import os, re, json, unicodedata
import numpy as np
from sentence_transformers import SentenceTransformer

files = sorted(f for f in os.listdir('corpus') if f.endswith('.md'))
names = [unicodedata.normalize('NFC', f[:-3]) for f in files]
docs  = [open(f'corpus/{f}', encoding='utf-8').read() for f in files]
model = SentenceTransformer('intfloat/multilingual-e5-small')

QSET10 = [
    ("칠러가 안 도는 것 같은데 챔버가 멈추는 기준이 뭔가요", 'SOP-ETCH-007_v2.1'),
    ("에치 리크 체크는 어떻게 하나요",                      'SOP-ETCH-006'),
    ("CVD 알람이 뜨면 뭘 먼저 하나요",                      'SOP-CVD-009'),
    ("He 배면압 설정 기준이 어떻게 되나요",                  'SOP-ETCH-012_v2.0'),
    ("포토 웨이퍼를 수동으로 넣는 절차",                     'SOP-포토-013'),
    ("시즈닝 런은 왜 도나요",                              'SOP-ETCH-005'),
    ("ESC 교체할 때 주의할 점",                            'SOP-ETCH-004'),
    ("파티클 카운트가 올라가면 어떻게 하나요",                 'SOP-ETCH-005'),
    ("가스라인 퍼지 순서",                                 'SOP-ETCH-003'),
    ("예방정비 주기가 어떻게 되나요",                        'SOP-ETCH-002'),
]
print(len(docs), '건 · 질문', len(QSET10), '개')
```

```python
# 미션 2 준비 (2) 예제 2-1 ~ 2-5 의 도구와 함수 · 코드는 예제와 같고 API 를 호출하지 않습니다
import subprocess                                    # 예제 2-1 (b)
from docx import Document                            # 예제 2-2 (d)
from openpyxl import load_workbook                   # 예제 2-2 (f)
import pandas as pd                                  # 예제 2-2 (g)

DROP = ['무단 반출 금지', '문서관리시스템', '문서번호']     # 예제 2-1 (c)

# 예제 2-3 (a)
def chunk_fixed(text, size=300, overlap=0):
    step = size - overlap
    return [text[i:i + size]
            for i in range(0, len(text), step)
            if text[i:i + size].strip()]

# 예제 2-3 (c)
def chunk_sent(text, target=300):
    sents = re.split(r'(?<=[.!?다])\s+', text)
    out, cur = [], ''
    for s in sents:
        if cur and len(cur) + len(s) > target:
            out.append(cur.strip()); cur = ''
        cur += s + ' '
    if cur.strip():
        out.append(cur.strip())
    return out

# 예제 2-3 (e)
def chunk_struct(text):
    title = text.split('\n', 1)[0].lstrip('# ').strip()
    parts, cur, head = [], [], None
    for ln in text.split('\n'):
        if ln.startswith('## '):
            if cur:
                parts.append((head, '\n'.join(cur)))
            head, cur = ln[3:].strip(), []
        else:
            cur.append(ln)
    if cur:
        parts.append((head, '\n'.join(cur)))
    return [f'[{title}] {h or ""}\n{b}'.strip() for h, b in parts if b.strip()]

# 예제 2-4 (a)
QE = model.encode(['query: ' + q for q, _ in QSET10], normalize_embeddings=True)

def build_chunks(fn):
    chunks, owner = [], []
    for nm, t in zip(names, docs):
        for c in fn(t):
            chunks.append(c); owner.append(nm)
    return chunks, owner

def evaluate(chunks, owner):
    CE = model.encode(['passage: ' + c for c in chunks],
                      normalize_embeddings=True, batch_size=64)
    h1 = h3 = 0
    for k, (q, gold) in enumerate(QSET10):
        order = (CE @ QE[k]).argsort()[::-1]
        tops = [owner[j] for j in order[:3]]
        h1 += tops[0] == gold
        h3 += gold in tops
    n = len(QSET10)
    return len(chunks), round(h1 / n * 100), round(h3 / n * 100)

# 예제 2-5 (d)
def dept_of(text):
    m = re.search(r'(?:작성|관리 부서|게시 부서): *([^\s·]+)', text)
    return m[1] if m else '-'

print('준비 완료')
```

```python
# 실습 미션 2 (a) 원본 10건 중 PDF · 워드 · MD 9건을 텍스트로 엽니다 (예제 2-1 (a) (b) · 2-2 (d) 와 같은 코드)
# 볼 것: 「pdf 4 · docx 4 · xlsx 1 · md 1」과 「9 건」. 엑셀은 (b) 에서 엽니다
RAW2 = 'docs_raw_m2'
for kind in ['pdf', 'docx', 'xlsx', 'md']:
    print(kind, len([f for f in os.listdir(f'{RAW2}/{kind}') if not f.startswith('.')]))

texts = {}
for f in sorted(os.listdir(f'{RAW2}/pdf')):
    if f.endswith('.pdf'):
        texts[f[:-4]] = subprocess.run(['pdftotext', f'{RAW2}/pdf/{f}', '-'],
                                       capture_output=True, text=True).stdout
for f in sorted(os.listdir(f'{RAW2}/docx')):
    if f.endswith('.docx'):
        paragraphs = [p.text for p in Document(f'{RAW2}/docx/{f}').paragraphs]
        texts[f[:-5]] = '\n'.join(paragraphs)
for f in sorted(os.listdir(f'{RAW2}/md')):
    if f.endswith('.md'):
        texts[f[:-3]] = open(f'{RAW2}/md/{f}', encoding='utf-8').read()
print(len(texts), '건 ·', list(texts))
```

```python
# 실습 미션 2 (b) 엑셀은 장식 줄을 건너뛰고 읽어 한 행을 한 문장으로 풉니다 (예제 2-2 (f) (g) (h))
# 볼 것: 「열 이름으로 읽힌 줄」이 코드 · 설비군 · 알람명인지, 「6 문장」과 문장 하나
XL2 = f'{RAW2}/xlsx/설비알람_코드표.xlsx'
ws = load_workbook(XL2).active
for row_no in range(1, 7):
    print(row_no, [c.value for c in ws[row_no]][:3])      # 몇째 줄이 열 이름인지 봅니다

# 건너뛸 줄 수: 열 이름(코드 · 설비군 · 알람명)이 있는 5행 위의 4줄
SKIP_ROWS = 4
df2 = pd.read_excel(XL2, skiprows=SKIP_ROWS)
print('열 이름으로 읽힌 줄:', list(df2.columns)[:3])

texts.pop('설비알람_코드표', None)        # 앞 실행에서 만든 문장은 지웁니다
# 열 이름 자리에 장식 줄이 들어가면 「코드」 열이 없어 문장을 만들 수 없습니다
if '코드' not in df2.columns:
    print('「코드」 열이 없어 문장을 만들지 못했습니다. 읽힌 표의 앞 3행:')
    print(df2.head(3).to_string())
else:
    sentences = []
    for _, r in df2.iterrows():
        sentences.append(f"알람 코드 {r['코드']} 는 {r['설비군']} 설비의 {r['알람명']} 알람이고, "
                         f"심각도는 {r['심각도']} · 1차 대응은 {r['1차 대응']} 입니다.")
    texts['설비알람_코드표'] = '\n'.join(sentences)
    print(len(sentences), '문장 ·', sentences[1])
```

```python
# 실습 미션 2 (c) 노이즈 단어를 더해 파일별 노이즈 비율을 측정합니다 (예제 2-1 (c) · 2-2 (a))
# 볼 것: 파일별 비율 표 10행. PDF 4건이 41~57 % 면 세 종류의 반복 줄이 모두 걸러진 것입니다
print(texts['NOTI-031'])                           # 쪽마다 되풀이되는 줄을 눈으로 확인합니다

# 예제 DROP 에 더할 단어: 보안 문구 줄 · 쪽 끝 「계속」 줄 · 출력 기록 줄
NEW_NOISE = ['기밀', '계속', 'Printed']
DROP2 = DROP + NEW_NOISE

def is_noise(line):                                # 단어가 들어 있거나 쪽번호(-1-)만 있는 줄
    page_no = line.strip().strip('-').strip()
    return any(word in line for word in DROP2) or (line.strip().startswith('-') and page_no.isdigit())

clean = {}
for name, text in texts.items():
    lines = [l for l in text.split('\n') if l.strip()]
    noise = [l for l in lines if is_noise(l)]
    clean[name] = '\n'.join(l for l in lines if not is_noise(l))
    print(f'{name:16s} {len(noise):2d} / {len(lines):2d} 줄 · {round(len(noise) / len(lines) * 100, 1)} %')
```

```python
# 실습 미션 2 (d) 두 크기로 잘라 질문 5개의 적중률을 비교하고 하나를 고릅니다 (예제 2-3 (a) · 2-4 (a))
# 볼 것: 두 크기의 청크 수 · top1 · top3 가 한 줄씩, 마지막 두 줄에 고른 크기와 NOTI-031 의 첫 청크
names, docs = list(clean), list(clean.values())         # 예제 2-4 (a) 의 build_chunks · evaluate 가 읽는 값
QSET10 = [('야간조 식당은 몇 시까지 하나요?',          'NOTI-031'),
          ('조별 인원 확인은 누가 하나요?',            'EHS-021'),
          ('가스 공급 밸브는 언제 차단하나요?',         'EHS-021'),
          ('에치 챔버 세정 SOP 의 세정액 대기 시간은?', 'SOP-ETCH-013'),
          ('E-102 알람이 뜨면 먼저 뭘 하나요?',        '설비알람_코드표')]
QE = model.encode(['query: ' + q for q, _ in QSET10], normalize_embeddings=True)

def by_size(size):                                 # 모든 문서를 size 글자씩 자릅니다 (예제 2-3 (a))
    def cut(text):
        return chunk_fixed(text, size)
    return build_chunks(cut)

# 비교할 청크 크기 두 개(글자 수)
SIZES = [100, 400]
result = {}
for size in SIZES:
    n, t1, t3 = evaluate(*by_size(size))
    result[size] = (t3, t1)
    print(f'{size:3d}자  청크 {n:3d}  top1 {t1:3d}%  top3 {t3:3d}%')

# 고른 크기: top3 가 높은 크기, top3 가 같으면 top1 이 높은 크기
BEST = SIZES[0]
for size in SIZES:
    if result[size] > result[BEST]:
        BEST = size
print('고른 크기', BEST, '자 · top3', result[BEST][0], '% · top1', result[BEST][1], '%')
print('NOTI-031 의 첫 청크:', chunk_fixed(clean['NOTI-031'], BEST)[0].replace('\n', ' / '))
```

```python
# 실습 미션 2 (e) 파일 이름에서 문서 유형 꼬리표를, 머리말에서 부서 꼬리표를 붙입니다 (예제 2-5 (c) · (d))
# 볼 것: 파일마다 유형 · 부서 · 청크 수 한 줄. 유형 규칙에 안 걸리는 파일은 「규칙에 안 걸림」
# ✏️ ======== 더 해 보기(선택) · 유형 꼬리표로 인정할 이름 ========
# 지금은 예제의 유형 네 개(SOP, QIR, SPEC, MTG)와 새로 더한 네 개(NOTI, EHS, MAN, EDU)가 들어 있습니다.
# 예제의 네 개만 남겨, 목록에 없는 유형의 파일이 「규칙에 안 걸림」으로 바뀌는지 확인합니다.
TYPES = ['SOP', 'QIR', 'SPEC', 'MTG', 'NOTI', 'EHS', 'MAN', 'EDU']
# ✏️ ======== 여기까지 ========
chunks_best, owner_best = by_size(BEST)            # (d) 에서 고른 크기로 자른 청크
for name, text in clean.items():
    kind = name.split('-')[0]                      # 파일 이름에서 첫 '-' 앞부분
    tag = kind if kind in TYPES else '규칙에 안 걸림'
    n_chunks = owner_best.count(name)              # 이 파일에서 나온 청크 수
    print(f'{name:16s} 유형 {tag:8s} · 부서 {dept_of(text):8s} · 청크 {n_chunks}')
print('청크 합계', len(chunks_best), f'({BEST}자로 자름)')
```

### 미션 2 · 바꿔 보고 답하기
**바꿔 보기 셀마다 값 하나를 고쳐 그 셀만 실행하고, 기준 결과와 비교합니다.**

바꿔 보기 셀은 그 값 하나만 바꿔 실습 미션 2 (b) ~ (e) 를 셀 안에서 다시 계산합니다. 다른 바꿔 보기 셀과 (a) ~ (e) 의 결과는 바뀌지 않으므로 값을 되돌리지 않아도 됩니다. 아래 도우미 셀을 먼저 한 번 실행합니다.

**바꿔 보기 1** 바꿔 보기 1 셀의 엑셀에서 건너뛸 줄 수 `SKIP_ROWS_1 = 4` 를 `0` 으로 고쳐 실행합니다. 「열 이름으로 읽힌 줄」과 100자 줄을 기준 결과와 비교합니다.

**바꿔 보기 2** 바꿔 보기 2 셀의 비교할 청크 크기 두 개 `SIZES_2 = [100, 400]` 을 다른 두 크기(예: `[200, 800]`)로 고쳐 실행합니다. 두 크기의 청크 수 · top1 · top3 를 기준 결과와 나란히 적어 둡니다.

**바꿔 보기 3** 바꿔 보기 3 셀의 노이즈로 더 지울 단어 `NEW_NOISE_3` 을 빈 목록 `[]` 으로 고쳐 실행합니다. PDF 4건의 노이즈 비율, 100자 줄, 「NOTI-031 의 첫 청크」 줄을 기준 결과와 비교합니다.

```python
# 실습 미션 2 바꿔 보기 도우미 · 값 하나만 바꿔 (b) ~ (e) 를 다시 계산하는 함수입니다 · 실행만 합니다
# (a) ~ (e) 가 만든 texts · clean · BEST 등은 바꾸지 않고, 함수 안에서만 계산합니다
assert 'BEST' in globals() and QSET10[0][1] == 'NOTI-031', '실습 미션 2 (a) ~ (e) 셀을 먼저 실행하세요'
BASE_SKIP_ROWS = 4                           # 실습 미션 2 (b) 의 값
BASE_NOISE = ['기밀', '계속', 'Printed']       # 실습 미션 2 (c) 의 값
BASE_SIZES = [100, 400]                      # 실습 미션 2 (d) 의 값
PDF4 = [f[:-4] for f in sorted(os.listdir(f'{RAW2}/pdf')) if f.endswith('.pdf')]
M2_QSET = list(QSET10)                       # 실습 미션 2 (d) 의 질문 5개
M2_QE = model.encode(['query: ' + q for q, _ in M2_QSET], normalize_embeddings=True)

def run_mission2(skip_rows=BASE_SKIP_ROWS, new_noise=BASE_NOISE, sizes=BASE_SIZES):
    texts_x = {k: v for k, v in texts.items() if k != '설비알람_코드표'}      # (a) 에서 연 9건
    table = pd.read_excel(XL2, skiprows=skip_rows)                           # (b) 엑셀
    header = list(table.columns)[:3]
    if '코드' in table.columns:
        texts_x['설비알람_코드표'] = '\n'.join(
            f"알람 코드 {r['코드']} 는 {r['설비군']} 설비의 {r['알람명']} 알람이고, "
            f"심각도는 {r['심각도']} · 1차 대응은 {r['1차 대응']} 입니다." for _, r in table.iterrows())
    drop = DROP + list(new_noise)                                            # (c) 노이즈
    def noisy(line):
        page_no = line.strip().strip('-').strip()
        return any(word in line for word in drop) or (line.strip().startswith('-') and page_no.isdigit())
    clean_x, ratio = {}, {}
    for name, text in texts_x.items():
        lines = [l for l in text.split('\n') if l.strip()]
        clean_x[name] = '\n'.join(l for l in lines if not noisy(l))
        ratio[name] = round(sum(noisy(l) for l in lines) / len(lines) * 100, 1)
    rows = {}                                                                # (d) 크기별 적중률
    for size in sizes:
        chunks_x, owner_x = [], []
        for name, text in clean_x.items():
            for c in chunk_fixed(text, size):
                chunks_x.append(c); owner_x.append(name)
        CE = model.encode(['passage: ' + c for c in chunks_x], normalize_embeddings=True, batch_size=64)
        h1 = h3 = 0
        for k, (q, gold) in enumerate(M2_QSET):
            tops = [owner_x[j] for j in (CE @ M2_QE[k]).argsort()[::-1][:3]]
            h1 += tops[0] == gold
            h3 += gold in tops
        rows[size] = (len(chunks_x), round(h1 / len(M2_QSET) * 100), round(h3 / len(M2_QSET) * 100))
    best = sizes[0]                          # top3 가 높은 크기, 같으면 top1 이 높은 크기
    for size in sizes:
        if (rows[size][2], rows[size][1]) > (rows[best][2], rows[best][1]):
            best = size
    first = chunk_fixed(clean_x['NOTI-031'], best)[0].replace('\n', ' / ')
    return dict(header=header, ratio=ratio, rows=rows, best=best, first=first, files=len(clean_x))

def show2(label, res):                       # 실습 미션 2 (b) ~ (e) 와 같은 값을 한 묶음으로 출력합니다
    print(f'[{label}]')
    print('  열 이름으로 읽힌 줄:', res['header'], '· 문서', res['files'], '건')
    print('  PDF 4건 노이즈 비율:', ' · '.join(f'{k} {res["ratio"][k]} %' for k in PDF4))
    for size, (n, t1, t3) in res['rows'].items():
        print(f'  {size:3d}자  청크 {n:3d}  top1 {t1:3d}%  top3 {t3:3d}%')
    print('  고른 크기', res['best'], '자 · 청크 합계', res['rows'][res['best']][0])
    print('  NOTI-031 의 첫 청크:', res['first'])

BASE_M2 = run_mission2()                     # 처음 값으로 계산한 기준 결과
show2('기준 · 처음 값', BASE_M2)
```

(1) 셀은 예제 3-1 (a) 조립 셀과 같고, 문서 묶음 두 줄(`CORPUS` · `MANIFEST`)만 재료 [1] 의 값(`'corpus_m3'` · `'corpus_m3_manifest.csv'`)으로 미리 바꿔 두었습니다. 고칠 곳 없이 그대로 실행합니다.
(2) 셀은 질문 `Q_D`(예제 3-3 (g))와 진단 함수 `diagnose()`(예제 3-4 (c)), `pd` 를 다시 만듭니다. (3) · (4) 셀은 예제 3-3 (g-) · (h) 와 같습니다.
미션 셀 안내의 「조립 셀(3-1 (a))」 · 「3-3 (g-) · (h) 셀」은 이 노트북에서는 (1) · (3) · (4) 셀을 가리킵니다.
(4) 셀과 실습 미션 3 (b) · (c) 는 Claude API 를 호출하므로 환경 준비 (2) · (3) 셀을 먼저 실행해 둡니다.

**볼 것** (3) 셀이 「이 묶음에는 QIR-2026-0712 「시정 조치」 절이 없어 순위 비교는 건너뜁니다」를 찍으면 새 코퍼스로 바뀐 것입니다. (4) 셀은 제목 접두 색인으로 바꾼 뒤 예제 질문 `Q_D`(ETCH-04 호기)를 한 번 묻습니다(API 1번). 이 묶음에는 ETCH-04 문서가 없으므로 「사내 문서에서 근거를 찾지 못했습니다」가 나오면 정상입니다.

```python
# 3-1 (a) 조립 셀: 파트 1 · 2 에서 만든 부품(청킹 · 꼬리표 · 인덱스 · 검색)과 답변 생성을 한 셀로 모읍니다
import os, re, csv, unicodedata
import numpy as np
from collections import namedtuple
from sentence_transformers import SentenceTransformer
import anthropic

CORPUS = 'corpus_m3'                   # 미션 3 새 문서 묶음 (예제 노트북 조립 셀은 'corpus')
MANIFEST = 'corpus_m3_manifest.csv'    # 미션 3 새 묶음의 목록 파일
model = SentenceTransformer('intfloat/multilingual-e5-small')
META = {unicodedata.normalize('NFC', r['file']): r
        for r in csv.DictReader(open(MANIFEST, encoding='utf-8-sig'))}

Chunk = namedtuple('Chunk', 'doc_id version type date section text')
chunks = []
for raw in sorted(os.listdir(CORPUS)):
    fn = unicodedata.normalize('NFC', raw)         # 지우면 한글 파일명에서 KeyError
    if not fn.endswith('.md'):
        continue
    m = META[fn]
    head, *secs = re.split(r'\n(?=## )', open(f'{CORPUS}/{raw}', encoding='utf-8').read())
    for s in [head] + secs:
        s = s.strip()
        if not s:                  # 머리 조각이 빈 문서는 건너뜀
            continue
        if secs and s == head.strip():
            s = '\n'.join(s.splitlines()[:8])          # 머리 청크는 제목·메타까지만
        chunks.append(Chunk(m['doc_id'], m['version'], m['type'], m['date'],
                            s.splitlines()[0].lstrip('# ').strip(), s[:1200]))

E = model.encode(['passage: ' + c.text for c in chunks],
                 normalize_embeddings=True, batch_size=64)

Hit = namedtuple('Hit', 'score doc_id version type date section text')

def search(q, k=4, exclude_types=()):
    qe = model.encode(['query: ' + q], normalize_embeddings=True)[0]
    cand = [i for i, c in enumerate(chunks) if c.type not in exclude_types]
    scores = E[cand] @ qe
    return [Hit(round(float(scores[j]), 3), *chunks[cand[j]])
            for j in scores.argsort()[::-1][:k]]

def format_ctx(hits):
    return '\n\n'.join(f'[{h.doc_id}_{h.version}] ({h.section})\n{h.text}'
                        for h in hits)

client = anthropic.Anthropic()          # ANTHROPIC_API_KEY 환경변수를 읽습니다
MODEL = 'claude-sonnet-5-5'

def llm(prompt, max_tokens=1500):                 # 글 한 덩어리를 보내고 답 글자만 돌려받습니다
    r = client.messages.create(                    # API 호출 한 번 = 요청 한 번(쓴 토큰만큼 요금)
        model=MODEL, max_tokens=max_tokens,        # 쓸 모델 · 답 길이 상한(토큰)
        output_config={'effort': 'low'},            # 짧은 질의응답이라 추론을 적게 씁니다
        messages=[{'role': 'user', 'content': prompt}])   # 보낼 대화: 사용자 메시지 하나
    if r.stop_reason == 'refusal':                 # 왜 멈췄는지: 거절이면 안내 문장으로
        return '[안내] 모델이 답변을 거절했습니다.'
    return ''.join(b.text for b in r.content if b.type == 'text').strip()   # 답 조각 중 글자 조각만 이어 붙입니다

TEMPLATE = """너는 SL-FAB 공정기술 담당자다.
아래 사내 문서 발췌만 근거로 답한다.

[발췌]
{ctx}

규칙: 문장마다 근거 [문서번호]를 붙인다.
발췌에 없으면 '사내 문서에서 근거를 찾지 못했습니다' 라고만 답한다.

질문: {q}"""

def ask(q, k=4, exclude_types=(), template=None):
    hits = search(q, k, exclude_types)
    ctx = format_ctx(hits)
    return llm((template or TEMPLATE).format(ctx=ctx, q=q)), hits

print('청크', len(chunks), '· 인덱스', E.shape, '· 모델', MODEL)
```

```python
# 미션 3 준비 (2) 질문 Q_D 와 진단 함수 · 코드는 예제와 같고 API 를 호출하지 않습니다
import pandas as pd                                  # 예제 3-4 (d)
Q_D = 'ETCH-04 호기 ESC 온도 드리프트 원인과 재발 방지 조치는?'      # 예제 3-3 (g) · 아래 (3) · (4) 셀이 씁니다

# 예제 3-4 (c)
NO_EVIDENCE = '근거를 찾지 못했습니다'

def diagnose(q, k=4, template=None):          # template 을 주면 그 프롬프트로 묻습니다(바꿔 보기 1)
    ans, hits = ask(q, k=k, template=template)
    docs_ = {h.doc_id for h in hits}
    vers = {(h.doc_id, h.version) for h in hits}
    row = {'질문': q[:26],
           '최고 점수': hits[0].score,
           '1위-4위 차': round(hits[0].score - hits[min(3, len(hits) - 1)].score, 3),
           '유형': '/'.join(h.type for h in hits),
           '같은 문서 여러 판': '있음' if len(vers) > len(docs_) else '없음',
           '출처 표기': '있음' if re.search(r'\[[A-Z]+-', ans) else '없음',
           '근거 없음': '예' if NO_EVIDENCE in ans else '아니오'}
    return row, ans

print('준비 완료')
```

```python
# 3-3 (g-) 원인 절은 지금 몇 위인가 (보조)
TITLE = {}
for c in chunks:
    TITLE.setdefault((c.doc_id, c.version), c.section)    # 첫 청크의 절 제목 = 문서 제목

def rank_of_section(index, doc_id, section):
    qe = model.encode(['query: ' + Q_D], normalize_embeddings=True)[0]
    sc = index @ qe
    for r, i in enumerate(sc.argsort()[::-1], 1):
        if chunks[i].doc_id == doc_id and chunks[i].section == section:
            return r, float(sc[i])

E_title = model.encode([f'passage: {TITLE[c.doc_id, c.version]} · {c.section}\n{c.text}'
                        for c in chunks], normalize_embeddings=True, batch_size=64)
if rank_of_section(E, 'QIR-2026-0712', '시정 조치') is None:     # 미션 3 코퍼스(corpus_m3)에는 이 문서가 없습니다
    print('이 묶음에는 QIR-2026-0712 「시정 조치」 절이 없어 순위 비교는 건너뜁니다 · 제목 접두 색인', E_title.shape)
else:
    r, s = rank_of_section(E, 'QIR-2026-0712', '시정 조치')
    print(f'제목 접두 전   {r}위 / {len(chunks)}   ({s:.3f})')
    r, s = rank_of_section(E_title, 'QIR-2026-0712', '시정 조치')
    print(f'제목 접두 후   {r}위 / {len(chunks)}   ({s:.3f})   ← 원인·조치가 함께 있는 절')
```

```python
# 3-3 (h) 제목 접두 색인으로 다시
assert len(E_title) == len(chunks), '3-3 (g-) 셀을 먼저 실행하세요 (조립 셀을 다시 돌렸으면 (g-) 도 다시)'
E = E_title                                   # 절 앞에 문서 제목을 붙여 만든 색인으로 교체
ans, hits = ask(Q_D)
for h in hits:
    print(f'{h.score:.3f}  {h.doc_id}_{h.version}  ({h.section[:20]})')
print(ans)
```

```python
# 미션 3 도우미 · 문서 폴더를 받아 청크와 색인을 만드는 load_index(), 발췌와 답을 함께 보는 show()
# 준비 (1) 조립 셀의 청킹과 준비 (3) · (4) 의 제목 접두 색인(예제 3-3 (g-) · (h))을 함수 하나로 묶은 것입니다. (c) 가 씁니다
def load_index(corpus, manifest):
    meta = {unicodedata.normalize('NFC', r['file']): r
            for r in csv.DictReader(open(manifest, encoding='utf-8-sig'))}
    out = []
    for raw in sorted(os.listdir(corpus)):
        fn = unicodedata.normalize('NFC', raw)
        if not fn.endswith('.md'):
            continue
        m = meta[fn]
        head, *secs = re.split(r'\n(?=## )', open(f'{corpus}/{raw}', encoding='utf-8').read())
        for s in [head] + secs:
            s = s.strip()
            if not s:                  # 머리 조각이 빈 문서는 건너뜀
                continue
            if secs and s == head.strip():
                s = '\n'.join(s.splitlines()[:8])
            out.append(Chunk(m['doc_id'], m['version'], m['type'], m['date'],
                             s.splitlines()[0].lstrip('# ').strip(), s[:1200]))
    title = {}
    for c in out:
        title.setdefault((c.doc_id, c.version), c.section)      # 첫 청크의 제목 = 문서 제목
    vecs = model.encode([f'passage: {title[c.doc_id, c.version]} · {c.section}\n{c.text}' for c in out],
                        normalize_embeddings=True, batch_size=64)
    print(corpus, '·', len({(c.doc_id, c.version) for c in out}), '건 · 청크', len(out), '· 인덱스', vecs.shape)
    return out, vecs

def show(q):                                       # 붙는 발췌와 답을 함께 봅니다
    ans, hits = ask(q)
    for h in hits:
        print(f'  {h.score:.3f}  {h.doc_id}_{h.version}  ({h.section[:20]})')
    print(' →', ans[:300].replace('\n', ' '))

print('load_index() · show() 준비 완료')
```

```python
# 실습 미션 3 (a) 새 문서 묶음으로 바뀌었는지 확인하고, 프롬프트의 근거 없음 규칙을 정합니다
# 볼 것: 「corpus_m3 · 12 건 · 청크 39 · 인덱스 (39, 384)」와 지금 프롬프트에 들어간 규칙
assert CORPUS == 'corpus_m3', '「미션 3 준비」 (1) ~ (4) 셀을 차례로 다시 실행하세요'
assert len(E) == len(chunks), '「미션 3 준비」 (1) ~ (4) 셀을 차례로 다시 실행하세요'
print(CORPUS, '·', len({(c.doc_id, c.version) for c in chunks}), '건 · 청크', len(chunks), '· 인덱스', E.shape)

RULE4 = "발췌에 없으면 '사내 문서에서 근거를 찾지 못했습니다' 라고만 답한다."

def make_template(use_rule):             # 프롬프트 네 칸(예제 3-2): 역할 · 발췌 · 출력 규칙 · 근거 없음(RULE4)
    return ('너는 SL-FAB 공정기술 담당자다.\n아래 사내 문서 발췌만 근거로 답한다.\n\n'
            '[발췌]\n{ctx}\n\n'
            '규칙: 문장마다 근거 [문서번호]를 붙인다.\n'
            + (RULE4 if use_rule else '') + '\n\n'
            '질문: {q}')

TEMPLATE = make_template(True)           # (b) · (c) 가 쓰는 기본 프롬프트 · 근거 없음 규칙을 넣습니다
print('근거 없음 규칙:', RULE4)
```

```python
# 실습 미션 3 (b) 재료 [2] 의 질문 다섯 개를 진단 함수로 실행합니다 (예제 3-4 (c) (d))
# 볼 것: 답이 있는 네 질문에 출처 [문서번호] 가 붙는지, 5번(답이 없는 질문)의 근거 없음 칸이 「예」인지
MY_Q = ['확산로 보트 로딩 속도 기준이 얼마인가요?',
        'DIFF-02 산화막 두께 산포가 커진 원인과 조치는?',
        'RCP-IMP-220 의 빔 전류 목표값과 관리 한계는?',
        '이온 소스를 교체한 뒤 에이징은 몇 분 하나요?',
        'DIFF-07 호기 PM 요일은 언제인가요?']            # 5번은 답이 없는 질문
rows = []
for q in MY_Q:
    row, ans = diagnose(q)
    rows.append(row)
    print(q, '\n →', ans[:160].replace('\n', ' '), '\n')
pd.set_option('display.width', 200)
print(pd.DataFrame(rows).to_string(index=False))
```

```python
# 실습 미션 3 (c) 구판이 섞이면 어떻게 되는지 보는 함수를 만들고, 기본(구판 없음)으로 한 번 실행합니다 (예제 3-3 (f))
# 볼 것: 발췌 네 칸의 「문서번호_판」, 진단 줄의 「같은 문서 여러 판」, 답의 허용 편차 값
import shutil
Q_V = '확산로 온도 프로파일 허용 편차는?'

def check_versions(add_old):
    """색인을 사본 폴더로 다시 만들어 Q_V 를 묻고, 구판을 넣었으면 구판 청크를 빼고 한 번 더 묻습니다.
    (b) 가 쓰던 색인(chunks · E)은 끝에 그대로 돌려놓습니다."""
    global chunks, E
    base_chunks, base_E = chunks, E          # (b) 가 쓰던 색인
    out = []
    try:                                     # 중간에 오류(API 오류 등)가 나도 finally 가 색인을 원래대로 둡니다
        folder = 'corpus_m3_check'           # corpus_m3/ 는 그대로 두고 사본 폴더에서 다룹니다
        shutil.rmtree(folder, ignore_errors=True)
        os.makedirs(folder)
        for fn in os.listdir(CORPUS):
            shutil.copy(f'{CORPUS}/{fn}', folder)
        if add_old:
            shutil.copy('corpus_m3_ext/SOP-DIFF-002_v1.0.md', folder)
        chunks, E = load_index(folder, MANIFEST)     # 「미션 3 준비」 (1) ~ (4) 와 같은 방식
        print('[지금 색인]'); show(Q_V)
        row = diagnose(Q_V)[0]; print(row); out.append(row)
        if add_old:                          # 구판이 섞였으면 옛 판 청크만 빼고 다시 묻습니다 (예제 3-3 (f))
            keep = []
            for i, c in enumerate(chunks):
                if not (c.doc_id == 'SOP-DIFF-002' and c.version == 'v1.0'):
                    keep.append(i)
            print('청크', len(chunks), '→', len(keep))
            chunks, E = [chunks[i] for i in keep], E[keep]
            print('[구판을 뺀 뒤]'); show(Q_V)
            row = diagnose(Q_V)[0]; print(row); out.append(row)
    finally:
        chunks, E = base_chunks, base_E
    return out

BASE_C = check_versions(add_old=False)       # 기본 결과 · 바꿔 보기 2 셀이 함께 출력합니다
```
