### API 키는 코드에 적지 않습니다

API 키는 API를 호출한 사람을 식별하는 문자열입니다. 사용 기록과 요금은 키별로 집계됩니다. 키를 가진 사람은 누구나 그 키로 호출할 수 있고, 요금은 키의 주인에게 청구됩니다.

코드에 적은 키는 파일을 공유하거나 저장소에 올리는 순간 함께 퍼집니다. 그래서 키는 환경 변수에 두고 코드에서는 환경 변수를 읽습니다.

> **주의**
>
> 키를 메신저, 화면 공유, 캡처 이미지에 노출하지 않습니다. 노출됐다면 바로 폐기하고 새로 발급받습니다.

### 환경 변수 설정

`llm.py`는 아래 환경 변수를 읽어 어느 API로 보낼지 고릅니다.

**Anthropic**

| 환경 변수 | 값 |
| --- | --- |
| `ANTHROPIC_API_KEY` | 발급받은 키 |

모델 이름은 설정하지 않아도 됩니다. 소형, 기본, 대형 모델 등급에 현행 Claude 모델이 기본값으로 들어 있습니다.

**사내 API(OpenAI 호환)**

| 환경 변수 | 값 |
| --- | --- |
| `LLM_PROVIDER` | `openai` |
| `OPENAI_BASE_URL` | 사내 API 주소. 보통 `/v1`로 끝납니다 |
| `OPENAI_API_KEY` | 사내 API 키 |
| `LLM_MODEL_SMALL` | 소형 모델 이름 |
| `LLM_MODEL_DEFAULT` | 기본 모델 이름 |
| `LLM_MODEL_LARGE` | 대형 모델 이름 |
| `LLM_STRUCTURED` | 구조화 출력을 요청하는 방식. 설정하지 않으면 `json_schema`입니다. 서버가 이 방식을 지원하지 않을 때만 바꿉니다(6장 "구조화 출력") |

**PC에서 모델을 내려받을 때(공개 LLM, 임베딩 모델, 리랭커)**

| 환경 변수 | 값 |
| --- | --- |
| `HF_ENDPOINT` | 사내 Hugging Face 프록시 주소. 비워 두면 huggingface.co에서 받습니다. 이 값도 Jupyter를 시작하기 전에 설정합니다 |
| `EMBED_MODEL` | 설정하지 않아도 됩니다. 다른 임베딩 모델을 쓰거나 모델 파일이 든 폴더를 따로 받았을 때, 모델 이름이나 폴더 경로를 넣습니다 |

사내에서 쓸 수 있는 모델이 하나뿐이면 세 변수에 같은 이름을 넣습니다. 그러면 모델 등급을 비교하는 예제와 실습에서는 차이가 나지 않습니다.

환경 변수는 Jupyter를 시작하기 전에 설정합니다. `setx`로 설정한 값은 그 뒤에 새로 연 창에만 적용됩니다. Windows PowerShell에서 다음과 같이 설정한 뒤 터미널을 새로 열고, 그 터미널에서 Jupyter를 시작합니다. VS Code로 노트북을 연다면 VS Code 창을 모두 닫았다가 다시 엽니다.

Anthropic API를 쓸 때:

```powershell
setx ANTHROPIC_API_KEY "발급받은_키"
```

사내 API를 쓸 때(위 표의 나머지 변수도 같은 방법으로 설정합니다):

```powershell

setx OPENAI_BASE_URL "사내_API_주소"
setx OPENAI_API_KEY "사내_API_키"
```

`setx`로 설정한 값은 모든 장의 노트북에 그대로 적용됩니다.

환경 변수를 미리 설정하지 못했다면 아래 "준비 (1)" 셀이 키, 주소, 모델 이름을 물어봅니다. 입력한 값은 화면에 보이지 않고 노트북 파일에도 저장되지 않습니다. 이 값은 커널이 살아 있는 동안만 쓰이므로, 노트북을 새로 열거나 커널을 다시 시작하면 다시 입력합니다. 단, 사내 API를 쓰려면 `LLM_PROVIDER=openai`만은 Jupyter를 시작하기 전에 설정해야 합니다. "준비 (1)" 셀이 이 값을 보고 어떤 값을 물을지 정하기 때문입니다. 이 값이 없으면 `llm.py`는 Anthropic API로 요청을 보냅니다.

### 연결 확인

아래 두 셀을 차례로 실행합니다. 두 셀이 오류 없이 끝나고 마지막 줄에 모델의 응답(보통 `준비 완료`)이 나오면 연결된 것입니다. 응답 문구는 모델에 따라 조금 다를 수 있습니다.

````python
# 준비 (1) 환경 변수가 비어 있으면 여기서 입력합니다
import os
from getpass import getpass

import llm

needed = (['ANTHROPIC_API_KEY'] if llm.provider() == 'anthropic'
          else ['OPENAI_BASE_URL', 'OPENAI_API_KEY', 'LLM_MODEL_SMALL', 'LLM_MODEL_DEFAULT',
                'LLM_MODEL_LARGE'])
for name in needed:
    if not os.environ.get(name):
        os.environ[name] = getpass(f'{name}: ').strip()
print('사용하는 API:', llm.provider())
````

````python
# 준비 (2) 공통으로 쓰는 것을 불러오고 연결을 확인합니다
import json
import time
from concurrent.futures import ThreadPoolExecutor

import pandas as pd

DATA = 'data'   # 재료 폴더. 장마다 쓰는 재료는 이 아래에 있습니다

# 설치: python -m pip install transformers torch
# 위 패키지는 예제 6-2만 씁니다
# 6장과 관통 프로젝트 2단계에서 씁니다
import csv
import random

with open(f'{DATA}/문의.csv', encoding='utf-8') as file:
    inquiries = list(csv.DictReader(file))        # 문의 60건. 열: 번호, 문의, 라벨, 긴급, 비고
LABELS = ['설비 이상', '품질', '안전', '문서와 데이터 요청']

# 100만 토큰당 단가(달러). 5장과 같습니다(Anthropic API의 기본 모델, 2026-11 확인)
PRICES = {'소형': {'input': 0.10, 'output': 0.50}, '기본': {'input': 2.00, 'output': 10.00},
          '대형': {'input': 4.00, 'output': 20.00}}


def cost(reply, tier):
    """응답 하나의 요금(달러). 입력 토큰과 출력 토큰(thinking 포함)에 각각의 단가를 곱해 더합니다"""
    price = PRICES[tier]
    return (reply.input_tokens * price['input'] + reply.output_tokens * price['output']) / 1_000_000


def parsed(text):
    """응답 텍스트를 JSON으로 파싱합니다. 실패하면 "파싱 오류"와 응답 앞부분을 돌려줍니다"""
    try:
        return json.loads(text)
    except json.JSONDecodeError:
        return f'파싱 오류: {text.strip()[:40]!r}'


def load_local_model():
    """예제 6-2용. 1장의 공개 LLM을 PC에서 불러옵니다. 처음 한 번 약 1GB를 내려받습니다"""
    import embed
    from transformers import AutoModelForCausalLM, AutoTokenizer
    from transformers.utils import logging as transformers_logging
    name = 'Qwen/Qwen2.5-0.5B-Instruct'
    embed.ask_endpoint(name)                     # 모델이 PC에 없고 HF_ENDPOINT가 비어 있으면 주소를 묻습니다
    transformers_logging.disable_progress_bar()
    return AutoTokenizer.from_pretrained(name), AutoModelForCausalLM.from_pretrained(name).eval()


print(f'문의 {len(inquiries)}건, 라벨 {LABELS}')


for alias in ['소형', '기본', '대형']:
    try:
        print(f'{alias} 모델:', llm.model_name(alias))
    except RuntimeError:
        print(f'{alias} 모델: 정하지 않음')

reply = llm.send([{'role': 'user', 'content': '"준비 완료"라고만 답해 주세요.'}], model='소형', max_tokens=200)
print(reply.text)
````

## 앞 단계의 코드: 1단계 풀이

````python
SYSTEM = '너는 반도체 후공정 라인의 문의에 답하는 도우미다. 사내 정보를 모르면 지어내지 말고 모른다고 답한다.'


def answer(question, tier='소형', tries=3, send=llm.send, sleep=time.sleep):
    """질문 하나에 답합니다. 돌려주는 dict의 키: question, status('완료', '확인 필요', '실패'),
    answer, input_tokens, output_tokens, cost, model(응답에 적힌 실제 모델 이름), reason"""
    result = {'question': question, 'status': '실패', 'answer': '', 'input_tokens': 0,
              'output_tokens': 0, 'cost': 0.0, 'model': '', 'reason': ''}
    for attempt in range(tries):
        try:
            reply = send([{'role': 'user', 'content': question}], system=SYSTEM, model=tier, max_tokens=2000)
        except llm.CallError as error:
            result['reason'] = str(error)
            if not error.retryable:
                return result                         # 요청을 고쳐야 풀리는 오류는 다시 보내지 않습니다
            if attempt < tries - 1:
                sleep(2 ** attempt + random.random())  # 1초, 2초 … 백오프에 지터를 더합니다
            continue
        result.update(answer=reply.text, input_tokens=reply.input_tokens, output_tokens=reply.output_tokens,
                      cost=cost(reply, tier), model=reply.model, reason=reply.stop_reason,
                      status='완료' if reply.stop_reason == 'end' else '확인 필요')
        return result
    return result


def mock_send(results):
    """미리 정해 둔 결과를 차례로 돌려주는 mock. 결과가 예외이면 발생시킵니다"""
    queue = list(results)

    def send(messages, **options):
        result = queue.pop(0)
        if isinstance(result, Exception):
            raise result
        return result
    return send


done = llm.Reply(text='완료된 답', input_tokens=50, output_tokens=20, stop_reason='end', model='mock', seconds=0.1)
cut = llm.Reply(text='잘린 답', input_tokens=50, output_tokens=2000, stop_reason='max_tokens', model='mock', seconds=0.1)
busy, missing = llm.CallError('rate_limit', 429, '요청이 많음'), llm.CallError('not_found', 404, '없는 모델')
no_wait = lambda seconds: None                       # 테스트에서는 기다리지 않습니다
print('429 두 번 뒤 성공:', answer('질문', send=mock_send([busy, busy, done]), sleep=no_wait)['status'])
print('429 세 번:', answer('질문', send=mock_send([busy, busy, busy]), sleep=no_wait)['status'])
print('404 한 번(다음 응답은 성공):', answer('질문', send=mock_send([missing, done]), sleep=no_wait)['status'])
cut_result = answer('질문', send=mock_send([cut]), sleep=no_wait)
print('잘린 응답:', cut_result['status'], cut_result)

results = [answer(item['문의']) for item in inquiries if item['번호'] in ('Q31', 'Q33', 'Q35', 'Q36', 'Q38')]
for result in results:
    print(f"\n[{result['status']}] {result['question']}\n  → {result['answer'].strip()[:100]}")
print(f"\n요금 합계 {sum(result['cost'] for result in results):.5f}달러, "
      f"실패와 확인 필요 목록 {[(result['question'], result['reason']) for result in results if result['status'] != '완료']}")
````

## 실습 P2. 답을 구조화된 데이터로 받기

교재의 절: "관통 프로젝트 2단계"

첫째 셀에서는 `ANSWER_SCHEMA`에 네 필드(`category`: 라벨 네 개 중 하나, `urgent`: 불리언, `needs_document`: 불리언, `answer`: 문자열)를 모두 필수로 적습니다. 업무 규칙은 두 가지입니다. 분류가 "문서와 데이터 요청"이면 `needs_document`가 참이어야 하고, 분류가 "안전"인데 `urgent`가 거짓이면 사람이 확인합니다. `answer_structured()`는 1단계의 `answer()`를 호출하되, `send` 인자에 스키마를 넣은 `llm.send`를 넘깁니다. 1단계의 상태가 "완료"인 응답만 파싱하고, `check()`로 업무 규칙 두 가지를 검사합니다. 규칙에 걸리면 "사람 확인", 걸리지 않으면 "자동 처리"입니다. 1단계에서 "확인 필요"였던 응답(잘림, 거절)은 다시 보내도 대부분 되풀이되므로 "사람 확인"으로 넘기고, 1단계에서 "실패"였으면 "실패"로 둡니다. 파싱하지 못한 응답도 예외를 발생시키지 않고 "실패"와 사유를 돌려줍니다. `LLM_STRUCTURED=prompt`로 실행하면 서버가 형식을 보장하지 않고, 둘째 셀처럼 병렬로 실행할 때 예외가 하나라도 발생하면 이미 받은 결과까지 잃기 때문입니다(5장). 문의 6건의 처리 상태와 사유가 출력되면 첫째 셀이 끝납니다. 첫째 셀은 API를 6번 호출합니다.

둘째 셀에서는 문의 12건을 한 번에 4건까지 병렬로 보내고 결과를 SQLite 파일 `P2_결과.db`의 `결과` 테이블에 저장합니다. 문의 번호가 기본 키이고, 이 장 "결과와 함께 남길 값"에 적은 값(실행 번호, 실제 모델 이름, 프롬프트 버전, 상태, 사유, 토큰, 요금)을 라벨과 함께 남깁니다. `save_result()`는 결과 하나를 upsert하고 바로 커밋합니다. `run_batch()`는 저장된 결과를 먼저 읽어 결과가 없거나 상태가 "실패"인 문의만 보내고, 결과를 꺼내는 대로 한 건씩 저장합니다. 아래 코드는 세 번 실행합니다. 1회는 8건째에서 멈췄다고 보고 앞 8건만 넘깁니다. 2회 전에는 Q21이 재시도 끝에 실패했다고 보고 그 행의 상태를 "실패"로 바꿉니다. 2회와 3회에 보낼 문의 수와 세 번 실행한 뒤의 행 수를 먼저 `my_guess`에 적습니다. 둘째 셀은 API를 13번(실제로 실패한 건이 있으면 더) 호출합니다.

완료 기준

- 첫째 셀: 6건의 처리 상태가 출력되고, 규칙에 걸린 건에는 사유가 나옵니다. 출력에 1단계에서 계산한 요금이 그대로 나옵니다.
- 둘째 셀: 2회에는 1회에 넘기지 않은 4건과 Q21만 보내고, 3회에는 아무것도 보내지 않습니다. 2회나 3회에 실제로 실패한 건이 있으면 그 건만 더 보냅니다.
- 둘째 셀: 세 번 실행한 뒤에도 `결과` 테이블의 행은 12개이고, Q21의 행은 실행 번호가 "2회"로 바뀝니다.
- 둘째 셀: 1회와 2회의 실행 시간이 요청 시간의 합보다 짧습니다.

힌트 1. `functools.partial(llm.send, schema=ANSWER_SCHEMA)`는 `schema`를 미리 채운 `llm.send`입니다. 이것을 `answer(question, send=...)`로 넘기면 재시도와 종료 이유 확인을 1단계 코드 그대로 씁니다.

힌트 2. upsert의 `DO UPDATE SET` 뒤에는 번호를 뺀 열마다 `열 = excluded.열`을 적습니다. `', '.join(f'{column} = excluded.{column}' for column in COLUMNS[1:])`로 만들 수 있습니다. 실패한 건은 `result['data']`가 `None`이므로 분류, 긴급, 문서, 답 열에는 `None`을 넣습니다.

힌트 3. `pool.map()`은 결과를 입력 순서대로 돌려주므로 `zip(todo, pool.map(...))`로 문의와 결과를 짝짓습니다. SQLite 연결은 만든 스레드에서만 쓸 수 있으므로, 저장은 병렬로 실행하는 함수 안이 아니라 결과를 꺼내는 반복문에서 합니다.

</details>

````python
import functools

ANSWER_SCHEMA = {'type': 'object', 'additionalProperties': False,
                 'required': ['category', 'urgent', 'needs_document', 'answer'],
                 'properties': {'category': {'type': 'string', 'enum': LABELS}, 'urgent': {'type': 'boolean'},
                                'needs_document': {'type': 'boolean'}, 'answer': {'type': 'string'}}}


def check(data):
    """업무 규칙 검사. 문제가 없으면 None, 있으면 사유 문자열을 돌려줍니다"""
    if data['category'] == '문서와 데이터 요청' and not data['needs_document']:
        return '문서 요청인데 사내 문서가 필요하지 않다고 답함'
    if data['category'] == '안전' and not data['urgent']:
        return '안전 문의인데 긴급이 아니라고 답함'
    return None


def answer_structured(question):
    """1단계의 answer()에 구조화 출력을 더합니다. 결과 dict에 data(파싱한 dict)와 status, reason을 넣습니다"""
    result = answer(question, send=functools.partial(llm.send, schema=ANSWER_SCHEMA))
    result['data'] = None
    if result['status'] == '확인 필요':               # 잘림과 거절은 다시 보내도 대부분 되풀이되므로 사람에게 넘깁니다
        result['status'], result['reason'] = '사람 확인', f"정상 종료가 아님({result['reason']})"
    if result['status'] != '완료':                    # '실패'는 그대로 둡니다
        return result
    try:
        result['data'] = json.loads(result['answer'])
        problem = check(result['data'])
    except (json.JSONDecodeError, KeyError, TypeError) as error:   # 서버가 형식을 보장하지 않을 때(LLM_STRUCTURED=prompt)
        result['data'], result['status'], result['reason'] = None, '실패', f'응답 형식 오류: {error!r}'
        return result
    result['status'], result['reason'] = ('사람 확인', problem) if problem else ('자동 처리', '')
    return result


for item in [item for item in inquiries if item['번호'] in ('Q21', 'Q24', 'Q31', 'Q43', 'Q49', 'Q56')]:
    result = answer_structured(item['문의'])
    print(f"[{result['status']}] {item['번호']} {item['문의']}\n  → {result['data']}, 사유: {result['reason']}, 요금 {result['cost']:.5f}달러")
````

````python
import sqlite3

PROMPT_VERSION = 'P2-v1'                          # SYSTEM이나 ANSWER_SCHEMA를 고치면 바꿉니다
COLUMNS = ['번호', '실행', '모델', '프롬프트', '상태', '사유', '분류', '긴급', '문서', '답', '입력', '출력', '요금']


def open_db(path):
    """결과 테이블이 든 SQLite 파일을 엽니다. 문의 번호가 기본 키입니다"""
    db = sqlite3.connect(path)
    db.execute(f'CREATE TABLE IF NOT EXISTS 결과 ({", ".join(COLUMNS)}, PRIMARY KEY (번호))')
    return db


def save_result(db, number, run_id, result):
    """결과 하나를 문의 번호를 기본 키로 upsert하고 바로 커밋합니다"""
    data = result['data'] or {}                   # 실패한 건은 data가 None입니다
    values = [number, run_id, result['model'], PROMPT_VERSION, result['status'], result['reason'],
              data.get('category'), data.get('urgent'), data.get('needs_document'), data.get('answer'),
              result['input_tokens'], result['output_tokens'], result['cost']]
    updates = ', '.join(f'{column} = excluded.{column}' for column in COLUMNS[1:])
    db.execute(f'INSERT INTO 결과 ({", ".join(COLUMNS)}) VALUES ({", ".join(["?"] * len(COLUMNS))}) '
               f'ON CONFLICT(번호) DO UPDATE SET {updates}', values)
    db.commit()                                   # 커밋해야 중간에 멈춰도 파일에 남습니다


def run_batch(db, items, run_id, workers=4):
    """저장된 결과가 없거나 상태가 '실패'인 문의만 병렬로 보내고, 결과를 꺼내는 대로 한 건씩 저장합니다.
    보낸 문의 수와 요청 시간(초)의 합을 돌려줍니다"""
    done = {number for (number,) in db.execute("SELECT 번호 FROM 결과 WHERE 상태 != '실패'")}
    todo = [item for item in items if item['번호'] not in done]

    def timed(item):                              # 병렬로 실행하는 함수. 요청 하나에 걸린 시간을 결과에 더합니다
        start = time.perf_counter()
        result = answer_structured(item['문의'])
        result['seconds'] = time.perf_counter() - start
        return result

    total = 0.0
    with ThreadPoolExecutor(max_workers=workers) as pool:
        for item, result in zip(todo, pool.map(timed, todo)):   # 저장은 연결을 만든 이 스레드에서 합니다
            save_result(db, item['번호'], run_id, result)
            total += result['seconds']
    return len(todo), total


my_guess = {'2회에 보낼 수': 5, '3회에 보낼 수': 0, '행 수': 12}
DB_PATH = 'P2_결과.db'
if os.path.exists(DB_PATH):
    os.remove(DB_PATH)                            # 실습을 처음부터 다시 할 수 있게 지난 실행의 파일을 지웁니다
db = open_db(DB_PATH)
numbers = ('Q02', 'Q05', 'Q12', 'Q16', 'Q21', 'Q22', 'Q29', 'Q34', 'Q38', 'Q44', 'Q51', 'Q55')
items = [item for item in inquiries if item['번호'] in numbers]
for run_id, batch in [('1회', items[:8]), ('2회', items), ('3회', items)]:   # 1회는 8건째에서 멈췄다고 봅니다
    if run_id == '2회':
        db.execute("UPDATE 결과 SET 상태 = '실패' WHERE 번호 = 'Q21'")   # Q21이 재시도 끝에 실패했다고 봅니다
    start = time.perf_counter()
    sent, seconds = run_batch(db, batch, run_id)
    print(f'{run_id}: 보낸 문의 {sent}건, 요청 시간의 합 {seconds:.1f}초, 실행 시간 {time.perf_counter() - start:.1f}초')
print('예측:', my_guess)
print('행 수:', db.execute('SELECT COUNT(*) FROM 결과').fetchone()[0])
for row in db.execute('SELECT 번호, 실행, 모델, 상태, 분류, 긴급, 문서, 사유 FROM 결과 ORDER BY 번호'):
    print(row)
````
